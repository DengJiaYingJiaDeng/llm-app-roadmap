```python
import numpy as np

def cosine_similarity(a:np.ndarray,b:np.ndarray) -> float:
    dot_product = np.dot(a,b)
    norm_a = np.linalg.norm(a)
    norm_b = np.linalg.norm(b)

    if norm_a == 0 or norm_b == 0:
        raise ValueError("zero vector has no cosine similarity")

    return float(dot_product / (norm_a * norm_b))



vector_a = np.array([1.0,2.0,3.0])
vector_b = np.array([1.0,2.0,3.0])
vector_c = np.array([-1.0,-2.0,-3.0])
vector_d = np.array([3.0,0.0,0.0])

print("a vs b:",cosine_similarity(vector_a,vector_b))
print("a vs c:",cosine_similarity(vector_a,vector_c))
print("a vs d:",cosine_similarity(vector_a,vector_d))

def normalize_vector(vector:np.ndarray) -> np.ndarray:
    norm = np.linalg.norm(vector)

    if norm == 0:
        raise ValueError("connot normalize a zero vector")

    return vector / norm

vector = np.array([3.0,4.0])
normalized = normalize_vector(vector)

print(vector)
print(normalized)
print(np.linalg.norm(normalized))
```
