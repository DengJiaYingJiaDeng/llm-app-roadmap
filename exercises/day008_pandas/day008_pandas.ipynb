```python
import pandas as pd

# =========================
# 0. 创建原始脏数据
# =========================

data = {
    "user_id": [101, 102, 103, 103, 105, 106, 107],
    "name": [
        " Alice ",
        "Bob",
        "charlie",
        "charlie",
        None,
        "EVE ",
        " frank"
    ],
    "age": [
        "23",
        "31",
        "27",
        "27",
        None,
        "unknown",
        "29"
    ],
    "city": [
        "Shanghai",
        "beijing",
        "Shanghai ",
        "Shanghai ",
        "Shenzhen",
        None,
        "BEIJING"
    ],
    "score": [
        88.5,
        92.0,
        76.5,
        76.5,
        None,
        90.5,
        81.0
    ]
}

df = pd.DataFrame(data)


# =========================
# 1. 记录原始行数
# =========================
original_rows = len(df)

print("原始行数：",original_rows)
# =========================
# 2. 统计原始缺失值
# =========================
print(df.isna().sum())

# =========================
# 3. 按 user_id 删除重复记录
# =========================
before_dedup = len(df)

df = df.drop_duplicates(
    subset=["user_id"]
)
after_dedup = len(df)

removed_duplicates = (
    before_dedup-after_dedup
)
print("\n去重前:", before_dedup)
print("去重后:", after_dedup)
print("删除重复行:", removed_duplicates)


# =========================
# 4. 清洗 name
# 去前后空格 + 首字母格式
# =========================

df["name"] =(df["name"].str.strip().str.title())

# =========================
# 5. 清洗 city
# 去前后空格 + 格式统一 + 填充缺失
# =========================
df["city"] = (df["city"].str.strip().str.title())

df["city"] = (df["city"].fillna("Unknown"))

# =========================
# 6. 清洗 age
# 非法字符串转为 NaN
# 再使用 nullable integer
# =========================
df["age"] = pd.to_numeric(df["age"],errors="coerce")

df["age"] = (df["age"].astype("Int64"))
# =========================
# 7. 删除 name 缺失的数据
# =========================
before_drop_name = len(df)

df = df.dropna(subset=["name"])

after_drop_name = len(df)

removed_missing_name = (before_drop_name-after_drop_name)

print("\n删除 name 缺失前:", before_drop_name)
print("删除 name 缺失后:", after_drop_name)
print("删除 name 缺失行:", removed_missing_name)

# =========================
# 8. score 缺失值保留
# 不进行处理
# =========================


# =========================
# 9. 最终数据检查
# =========================
print("\n最终数据:")
print(df)

print("\n最终 shape:")
print(df.shape)

print("\n最终 dtypes:")
print(df.dtypes)

print("\n最终缺失值:")
print(df.isna().sum())

# =========================
# 10. 创建数据质量报告
# =========================
quality_report = pd.DataFrame({
    "column": df.columns,
    "dtype": df.dtypes.astype(str).values,
    "missing_count": df.isna().sum().values,
    "missing_rate": (
        df.isna().mean().values * 100
    ),
    "unique_count": [
        df[col].nunique(dropna=True)
        for col in df.columns
    ]
})

print("\n数据质量报告:")
print(quality_report)
```
